```python
# 1) sum of array

n=[1,2,3,4,5]
sum=0
for i in range(len(n)):
    
    sum +=n[i]
print("sum of array  :",sum)
```

```python
#2) find the max element in array
num=[1,2,3,4,5,3,2]
max=num[0]
for i in range(len(num)):
    if (max<=num[i]):
        max=num[i]
print("max of array:",max)
```

```python
#3) count even number in array
n=[1,2,7,3,5,6]
count=0
for i in n:
    if (i%2==0):
        count+=1
print("number of even =",count)
```

```python
#4) reverse an array

num=[1,2,3,4,5]
rev=[]

for n in num:
    rev.insert(0,n)
print(" reverse of array =",rev )
```

```python
#5) check palindromic
stg=list("madam")
rev=[]
for n in stg:
    rev.insert(0,n)
print("reverse a string=",rev)

if stg==rev:
    print("the given string is palindromic")
else:
    print("the given string is not palindromic")
```

```python
#6) sum of digits
n=1234
sum=0
while n>0:
    LD = n%10
    n = n//10
    sum += LD
print(sum)
```

```python
#7)count vowels in given string
num="dinesh karthick"
count=0
for n in num:
    if n.lower() in "aeiou":
        count=count+1
print("vowels count=",count)
```

```python
#8)second largest element in array
n=[5,4,3,2,1]
SA=sorted(n)
print(SA)
print(SA[-2])
```

```python
#8)second largest element in array
n=[1,3,5,4,3,6,2,1]

max=n[0]
max2=n[0]
for num in n:
    if max < num :
        max2=max
        max=num
    elif num>max2 and num != max:
        max2=num
print(max2)
```

```python
#9)  factorial numbers
n=5
fact=1
for i in range(1,n+1):
    fact=fact*i
print("factorial of numbers =",fact)
```

```python
#10)check prime numbers for given number
n=97
prime=0
for i in range (1,n+1):
    if n%i==0:
        prime=prime+1
if prime==2:
    print("the given",n,"is prime")
else:
    print("the given",n,"is not prime")
```
